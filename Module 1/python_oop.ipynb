Viết class và cài phương thức softmax


```python
import torch
import torch.nn as nn
```

```python
#Kieu du lieu cua torch la tensor
data =  torch.Tensor([1, 2, 3])
data
```

```python
softmax_function = nn.Softmax(dim=0) #chieu = 0
output = softmax_function(data)
output
```

```python
class MySoftMax(nn.Module):
    def __init__(self):
        super().__init__()
    
    def forward(self, x):
        x_exp = torch.exp(x)
        total = x_exp.sum(0, keepdims = True)
        return x_exp/total

my_softmax = MySoftMax()
output = my_softmax(data)
output
```

```python
data = torch.Tensor([1, 2, 30000000])
data
```

```python
#softmax binh thuong se bi NaN khi gap gia tri lon
output = my_softmax(data)
output 
```

```python
#Stable softmax
class StableSoftMax(nn.Module):
    def __init__(self):
        super().__init__()
    
    def forward(self, x):
        x_max = torch.max(x, dim=0, keepdims=True)
        x_exp = torch.exp(x - x_max.values)
        partition = x_exp.sum(0, keepdims=True)
        return x_exp / partition
    
data = torch.Tensor([1, 2, 3000000])
softmax_stable = StableSoftMax()
output = softmax_stable(data)
output
    
```

```python
from abc import ABC, abstractmethod

class Person(ABC):
    def __init__(self, name: str, yob:int):
        self._name = name
        self._yob = yob

    def get_yob(self):
        return self._yob

    @abstractmethod
    def describe(self):
        pass
```

```python
class Student(Person):
    def __init__(self, name, yob, grade):
        super().__init__(name = name, yob = yob)
        self.__grade = grade
    def describe(self):
        print(f"Student name: {self._name} - Student nam sinh: {self._yob} - Grade: {self.__grade}")
        
```

```python
class Doctor(Person):
    def __init__(self, name, yob, specialist: str):
        super().__init__(name = name, yob = yob)
        self.__specialts = specialist

    def describe(self):
        print(f"Doc name: {self._name} - Student nam sinh: {self._yob} - Specialits: {self.__specialts}")
```

```python
doctor1 = Doctor('Banh', 1992, 'Tim')
doctor2 = Doctor('Toan', 1994, 'Nguc')

doctor1.describe()
doctor2.describe()
```

```python
student1 = Student('Thai', 1999, 10)
student1.describe()
```

```python
class Teacher(Person):
    def __init__(self, name, yob, subject:str):
        super().__init__(name = name, yob = yob)
        self.__subject = subject
    def describe(self):
        print(f"name: {self._name} - yob: {self._yob} - subject =  {self.__subject}")
```

```python
teacher1 = Teacher('Lam', 1999, 'Toan')
teacher1.describe()
```

```python
class Ward:
    def __init__(self, name):
        self._name = name
        self._list_people = list()

    def add_person(self, person: Person):
        self._list_people.append(person)

    def describe(self):
        print(f"Ten ward = {self._name}")
        for p in self._list_people:
            p.describe()
    
    def count_doctor(self):
        counter = 0
        for p in self._list_people:
            if isinstance(p, Doctor):
                counter += 1
        return counter
    def sort_age(self):
        self._list_people.sort(key=lambda x: x.get_yob(), reverse=True)

    def compute_average_yob(self):
        total = 0 
        counter = 0
        for p in self._list_people:
            if isinstance(p, Doctor):
                counter += 1
                total += p.get_yob()
        return total/counter
            
```

```python
ward1 = Ward('Hai')
```

```python
ward1.add_person(student1)
ward1.describe()
```

```python
ward1.add_person(doctor1)
ward1.add_person(doctor2)
ward1.describe()
```

```python
ward1.count_doctor()
```

```python
ward1.sort_age()
ward1.describe()
```

```python
ward1.compute_average_yob()
```

Exercise 3: xây dựng class Stack

```python
class MyStack:
    def __init__(self, capacity):
        self.__capacity = capacity
        self.__stack = []

    #tra ve True khi len = 0    
    def is_empty(self):
        return len(self.__stack) == 0
    
    def is_full(self):
        return len(self.__stack) == self.__capacity
    
    def pop(self):
        if self.is_empty():
            print("stack rong")
            return None
        else:
            return self.__stack.pop() #ham pop cua list tu dong lay phan tu cuoi
    
    def push(self, value):
        if self.is_full():
            print("stack day")
            return None
        else:
            self.__stack.append(value)
    
    def top(self):
        if self.is_empty():
            print("stack rong")
            return None
        else: 
            return self.__stack[-1]
```

```python
stack1 = MyStack(3)
stack1.pop()
stack1.push(1)
stack1.push(3)
stack1.push(5)
stack1.pop()
stack1.top()
```

Exercise 4: xây dựng class Queue


```python
class MyQueue:
    def __init__(self, capacity: int):
        self.__capacity = capacity
        self.__queue = []
    
    def is_empty(self):
        return len(self.__queue) == 0
    
    def is_full(self):
        return len(self.__queue) == self.__capacity
    
    def dequeue(self):
        if self.is_empty():
            print("Queue rong")
            return None
        else:
            return self.__queue.pop(0)
    
    def enqueue(self, value):
        if self.is_full():
            self.__queue.pop(0)
            self.__queue.append(value)
        else:
            self.__queue.append(value)

    def front(self):
        if self.is_empty():
            print("queue rong")
            return None
        else:
            return self.__queue[0]
    
    def __str__(self):
        # Trả về một chuỗi văn bản (String) chứa cái mảng nội bộ
        return f"Trạng thái Queue hiện tại: {self.__queue}"
```

```python
queue1 = MyQueue(3)
queue1.dequeue()


```

```python
queue1.enqueue(1)
queue1.enqueue(2)
queue1.enqueue(3)
queue1.enqueue(4)
print(queue1)
queue1.front()
```
