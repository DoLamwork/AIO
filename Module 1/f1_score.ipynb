Exercise 1: Viết function đánh giá classificcation model bằng f1-score

```python
def calc_f1_score(tp, fp, fn):
    if type(tp) != int or type(fp) != int or type(fn) != int:
        raise ValueError("tp, fp, and fn must be integers")
    if tp < 0 or fp < 0 or fn < 0:
        raise ValueError("tp, fp, and fn must be non-negative integers")
    precision = tp / (tp + fp) if (tp + fp) > 0 else 0
    recall = tp / (tp + fn) if (tp + fn) > 0 else 0
    f1_score = 2 * (precision * recall) / (precision + recall) if (precision + recall) > 0 else 0
    return f1_score
```

```python
calc_f1_score(10, 5, 3)
```

```python
calc_f1_score(0, -2, -4)
```

```python
import math
def is_number(n):
    try: 
        float(n)
        return True
    except ValueError:   
        return False
assert is_number("3.14") == True
assert is_number("abc") == False
print(is_number("3.14"))
print(is_number("abc"))
```

```python
import math
def calc_sig(x):
    return 1 / (1 + math.exp(-x))
assert calc_sig(0) == 0.5
```

```python
def calc_elu(x, alpha=0.01):
    return x if x > 0 else alpha * (math.exp(x) - 1)
assert calc_elu(1) == 1
assert calc_elu(-1) == 0.01 * (math.exp(-1) - 1)
```

```python
def calc_relu(x):
    return x if x > 0 else 0
assert calc_relu(1) == 1
assert calc_relu(-1) == 0
```
